```python
# ================================= Imports =================================
import numpy as np
import pandas as pd
import os
import random
import glob
from matplotlib import pyplot as plt
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
import torch.optim as optim
from torchinfo import summary
import types

# ================================= models.py =================================
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        self.input_size = input_size[0] * input_size[1] # (7,7) -> 49
        self.input_layer = nn.Linear(self.input_size, hidden_size)
        self.relu = nn.ReLU()
        self.hidden_layers = nn.Sequential(
            nn.Linear(hidden_size, int(hidden_size)),
            nn.ReLU(),
            nn.Linear(hidden_size, int(hidden_size)),
            nn.ReLU(),
        )
        self.output_layer = nn.Linear(int(hidden_size), output_size)

    def forward(self, x):
        x = x.view(-1, self.input_size) # (batch_size, 7, 7) -> (batch_size, 49)
        x = self.input_layer(x)
        x = self.hidden_layers(x)
        x = self.output_layer(x)
        return x

# ================================= dataloader.py =================================
class CustomDataset(Dataset):
    # feature_indices 인자 추가
    def __init__(self, flag='train', feature_indices=None):
        dataset = np.load('/content/drive/MyDrive/ace_level2_98_11/kasi_DL/full_dataset.npz', allow_pickle=True)

        inp = dataset['inp']
        tar = dataset['tar']
        tar_time = dataset['time']

        # 데이터 슬라이싱 로직 추가
        if feature_indices is not None:
            print(f"선택된 피처 인덱스: {feature_indices}")
            inp = inp[:, :, feature_indices]

        valid_idx = sorted(list(set(range(len(tar))) - set(np.where(np.isnan(tar))[0])))
        inp, tar, tar_time = inp[valid_idx], tar[valid_idx], tar_time[valid_idx]
        inp, tar = self._data_preprocess(inp, tar)
        month = np.array([t.month for t in pd.to_datetime(tar_time)])

        if flag == 'train':
            mask = month < 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        elif flag == 'valid':
            mask = month == 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        elif flag == 'test':
            mask = month > 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        else:
            raise NotImplementedError

    def __len__(self):
        return len(self.tar)

    def __getitem__(self, idx):
        return torch.tensor(self.inp[idx], dtype=torch.float32), torch.tensor(self.tar[idx], dtype=torch.float32)

    def _data_preprocess(self, inp, tar):
        original_shape = inp.shape
        flat = inp.reshape(original_shape[0] * original_shape[1], original_shape[2])
        df = pd.DataFrame(flat)
        df_interp = df.interpolate(method='linear', axis=0).bfill().ffill()
        inp_interp = df_interp.values.reshape(original_shape)
        inp_max = np.max(inp_interp, axis=(0, 1))
        x = inp_interp / inp_max
        tar_max = np.max(tar, axis=0)
        y = tar / tar_max
        return x, y

# ================================= exp.py =================================
class Exp:
    def __init__(self, args):
        super(Exp, self).__init__()
        self.args = args
        random.seed(self.args.seed)
        torch.manual_seed(self.args.seed)

        # 피처 개수를 args.features 리스트 길이로
        self.args.inp_features = len(self.args.features)
        print(f"입력 피처 개수: {self.args.inp_features}")

        if torch.cuda.is_available():
            self.device = torch.device('cuda')
        else:
            self.device = torch.device("cpu")
            print("현재 GPU를 사용할 수 없습니다.")

        self.model_path = os.path.join('models', self.args.ex_name, self.args.model_name)
        os.makedirs(self.model_path, exist_ok=True)

        self.model = SimpleNN(input_size=(self.args.inp_seq, self.args.inp_features),
                              hidden_size=self.args.hidden_size,
                              output_size=self.args.tar_seq).to(self.device)

    def prepare_data(self):
        # CustomDataset에 feature_indices를 전달
        if self.args.flag == 'train':
            train_dataset = CustomDataset(flag='train', feature_indices=self.args.features)
            val_dataset = CustomDataset(flag='valid', feature_indices=self.args.features)
            return train_dataset, val_dataset
        else:
            test_dataset = CustomDataset(flag='test', feature_indices=self.args.features)
            return test_dataset

    # 동일하게
    def model_train(self):
        train_dataset, val_dataset = self.prepare_data()
        train_loader = DataLoader(train_dataset, batch_size=self.args.batch_size, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=self.args.val_batch_size, shuffle=False)
        model = self.model
        summary(model, input_size=(self.args.batch_size, self.args.inp_seq, self.args.inp_features))
        criterion = nn.MSELoss()
        optimizer = optim.Adam(model.parameters(), lr=self.args.lr)
        train_loss_history, val_loss_history, best_val_loss = [], [], float('inf')

        for epoch in range(1, self.args.epochs + 1):
            model.train()
            running_loss = 0.0
            for x, y in train_loader:
                x, y = x.to(self.device), y.to(self.device)
                outputs = model(x)
                loss = criterion(outputs, y)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                running_loss += loss.item()
            epoch_loss = running_loss / len(train_loader)
            train_loss_history.append(epoch_loss)
            current_val_loss = self.validate(model, val_loader, criterion)
            val_loss_history.append(current_val_loss)
            print(f"Epoch [{epoch}/{self.args.epochs}] | Train Loss: {epoch_loss:.5f} | Val Loss: {current_val_loss:.5f}")
            if current_val_loss < best_val_loss:
                best_val_loss = current_val_loss
                torch.save(model.state_dict(), os.path.join(self.model_path, 'best_model_state_dict.pth'))
                print(f"Model saved at epoch {epoch} with val loss: {best_val_loss:.5f}")

        plt.plot(train_loss_history, label='train')
        plt.plot(val_loss_history, label='val')
        plt.legend()
        plt.show()

    def validate(self, model, val_loader, criterion):
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for x, y in val_loader:
                x, y = x.to(self.device), y.to(self.device)
                outputs = model(x)
                loss = criterion(outputs, y)
                val_loss += loss.item()
        return val_loss / len(val_loader)
```

```python
# ================================= Imports =================================
import numpy as np
import pandas as pd
import os
import random
import glob
from matplotlib import pyplot as plt
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
import torch.optim as optim
from torchinfo import summary
import types

# ================================= models.py 내용 =================================
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        self.input_size = input_size[0] * input_size[1]
        self.input_layer = nn.Linear(self.input_size, hidden_size)
        self.relu = nn.ReLU()
        self.hidden_layers = nn.Sequential(
            nn.Linear(hidden_size, int(hidden_size)),
            nn.ReLU(),
            nn.Linear(hidden_size, int(hidden_size)),
            nn.ReLU(),
        )
        self.output_layer = nn.Linear(int(hidden_size), output_size)

    def forward(self, x):
        x = x.view(-1, self.input_size)
        x = self.input_layer(x)
        x = self.hidden_layers(x)
        x = self.output_layer(x)
        return x

# ================================= dataloader.py 내용 =================================
class CustomDataset(Dataset):
    def __init__(self, flag='train', feature_indices=None):
        dataset_path = '/content/drive/MyDrive/ace_level2_98_11/kasi_DL/full_dataset.npz'
        dataset = np.load(dataset_path, allow_pickle=True)

        inp = dataset['inp']
        tar = dataset['tar']
        tar_time = dataset['time']

        if feature_indices is not None:
            print(f"선택된 피처 인덱스: {feature_indices}")
            inp = inp[:, :, feature_indices]

        valid_idx = sorted(list(set(range(len(tar))) - set(np.where(np.isnan(tar))[0])))
        inp, tar, tar_time = inp[valid_idx], tar[valid_idx], tar_time[valid_idx]
        inp, tar = self._data_preprocess(inp, tar)
        month = np.array([t.month for t in pd.to_datetime(tar_time)])

        if flag == 'train':
            mask = month < 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        elif flag == 'valid':
            mask = month == 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        elif flag == 'test':
            mask = month > 9
            self.inp, self.tar, self.tar_time = inp[mask], tar[mask], tar_time[mask]
        else:
            raise NotImplementedError

    def __len__(self):
        return len(self.tar)

    def __getitem__(self, idx):
        return torch.tensor(self.inp[idx], dtype=torch.float32), torch.tensor(self.tar[idx], dtype=torch.float32)

    def _data_preprocess(self, inp, tar):
        original_shape = inp.shape
        flat = inp.reshape(original_shape[0] * original_shape[1], original_shape[2])
        df = pd.DataFrame(flat)
        df_interp = df.interpolate(method='linear', axis=0).bfill().ffill()
        inp_interp = df_interp.values.reshape(original_shape)
        inp_max = np.max(inp_interp, axis=(0, 1))
        x = inp_interp / inp_max
        tar_max = np.max(tar, axis=0)
        y = tar / tar_max
        return x, y

# ================================= exp.py 내용 =================================
class Exp:
    def __init__(self, args):
        super(Exp, self).__init__()
        self.args = args
        random.seed(self.args.seed)
        torch.manual_seed(self.args.seed)

        self.args.inp_features = len(self.args.features)
        print(f"입력 피처 개수: {self.args.inp_features}")

        if torch.cuda.is_available():
            self.device = torch.device('cuda')
            print(f"GPU: {torch.cuda.get_device_name(0)}")
        else:
            self.device = torch.device("cpu")
            print("현재 GPU를 사용할 수 없습니다.")

        # test 모드에서도 올바른 모델 경로를 찾도록 수정
        self.model_path = os.path.join('re_pys/models', self.args.ex_name, self.args.model_name)
        os.makedirs(self.model_path, exist_ok=True)

        self.model = SimpleNN(input_size=(self.args.inp_seq, self.args.inp_features),
                              hidden_size=self.args.hidden_size,
                              output_size=self.args.tar_seq).to(self.device)

    def prepare_data(self):
        if self.args.flag == 'train':
            train_dataset = CustomDataset(flag='train', feature_indices=self.args.features)
            val_dataset = CustomDataset(flag='valid', feature_indices=self.args.features)
            return train_dataset, val_dataset
        else: # 'test'
            test_dataset = CustomDataset(flag='test', feature_indices=self.args.features)
            return test_dataset

    def model_train(self):
        train_dataset, val_dataset = self.prepare_data()
        train_loader = DataLoader(train_dataset, batch_size=self.args.batch_size, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=self.args.val_batch_size, shuffle=False)
        model = self.model
        summary(model, input_size=(self.args.batch_size, self.args.inp_seq, self.args.inp_features))
        criterion = nn.MSELoss()
        optimizer = optim.Adam(model.parameters(), lr=self.args.lr)
        train_loss_history, val_loss_history, best_val_loss = [], [], float('inf')

        for epoch in range(1, self.args.epochs + 1):
            model.train()
            running_loss = 0.0
            for x, y in train_loader:
                x, y = x.to(self.device), y.to(self.device)
                outputs = model(x)
                loss = criterion(outputs, y)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                running_loss += loss.item()
            epoch_loss = running_loss / len(train_loader)
            train_loss_history.append(epoch_loss)
            current_val_loss = self.validate(model, val_loader, criterion)
            val_loss_history.append(current_val_loss)
            print(f"Epoch [{epoch}/{self.args.epochs}] | Train Loss: {epoch_loss:.5f} | Val Loss: {current_val_loss:.5f}")
            if current_val_loss < best_val_loss:
                best_val_loss = current_val_loss
                torch.save(model.state_dict(), os.path.join(self.model_path, 'best_model_state_dict.pth'))
                print(f"Model saved at epoch {epoch} with val loss: {best_val_loss:.5f}")

        plt.figure(figsize=(8, 5))
        plt.plot(train_loss_history, label='train_loss')
        plt.plot(val_loss_history, label='val_loss')
        plt.legend(loc='upper right')
        plt.title(f"Training and Validation Loss for features {self.args.features}")
        plt.show()

    def validate(self, model, val_loader, criterion):
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for x, y in val_loader:
                x, y = x.to(self.device), y.to(self.device)
                outputs = model(x)
                loss = criterion(outputs, y)
                val_loss += loss.item()
        return val_loss / len(val_loader)

    #######################333## 추가/수정 ###########################################3
    def model_test(self):
        test_dataset = self.prepare_data()
        test_loader = DataLoader(test_dataset, batch_size=self.args.val_batch_size, shuffle=False)

        load_path = os.path.join(self.model_path, "best_model_state_dict.pth")
        model = self.model
        model.load_state_dict(torch.load(load_path, map_location=self.device))
        model.to(self.device)
        model.eval()
        print(f"Best model loaded from {load_path}")

        criterion = nn.MSELoss()
        total_loss = 0.0
        predictions, targets = [], []

        with torch.no_grad():
            for x, y in test_loader:
                x, y = x.to(self.device), y.to(self.device)
                outputs = model(x)
                loss = criterion(outputs, y)
                total_loss += loss.item()

                predictions.append(outputs.detach().cpu().numpy())
                targets.append(y.detach().cpu().numpy())

        avg_loss = total_loss / len(test_loader)
        print(f"\n===== 최종 테스트 성능 =====")
        print(f"Test Set MSE Loss: {avg_loss:.6f}")
        print("==========================")

        # 예측 결과 시각화
        predictions = np.concatenate(predictions, axis=0)
        targets = np.concatenate(targets, axis=0)

        target_cols = ['C6to5', 'O7to6', 'FetoO']
        fig, axes = plt.subplots(self.args.tar_seq, 1, figsize=(15, 10), sharex=True)
        fig.suptitle(f'Test Set: Predictions vs. Actuals for features {self.args.features}', fontsize=16)

        for i in range(self.args.tar_seq):
            axes[i].plot(targets[:, i], label='Actual')
            axes[i].plot(predictions[:, i], label='Predicted', linestyle='--')
            axes[i].set_title(target_cols[i])
            axes[i].legend()

        plt.xlabel('Time Step (in test set)')
        plt.tight_layout(rect=[0, 0.03, 1, 0.95])
        plt.show()
```

```python
# train & test

config = types.SimpleNamespace()

config.features = [0, 1, 2, 3, 4, 5, 6]

# train 실행 후, test로 변경해야 합니다.
config.flag = 'test'

config.inp_seq = 7
config.tar_seq = 3
config.hidden_size = 128
config.batch_size = 32
config.val_batch_size = 1
config.lr = 0.0005
config.epochs = 40
config.seed = 777
config.ex_name = f'notebook_features_{"_".join(map(str, config.features))}'
config.model_name = 'MLP'

print(f"실험 '{config.ex_name}'을 시작합니다. (모드: {config.flag})")
exp_instance = Exp(config)

if config.flag == 'train':
    exp_instance.model_train()
    print("\n 훈련이 완료되었습니다. 이제 flag를 'test'로 바꾸고 최종 성능을 평가할 수 있습니다.")
else:
    exp_instance.model_test()
```
